### Домашні завдання на відпрацювання поняття функції та її похідної

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.symbols('x')

func = 10 * x - 5
derivative_func = sp.diff(func, x)

print("Похідна лінійної функції 10x-5:", derivative_func)
```

```python
xs = np.linspace(0,10,10)
func = 10 * xs - 5
dfunc = 10 * np.ones_like(xs)

plt.figure(figsize=(10, 5))
plt.plot(xs, func, label = 'f(x) = 10x-5', color='blue')
plt.plot(xs, dfunc, label="f'(x) = 10", color='green')

plt.title("Графік функції $f(x) = 10x - 5$ та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

**Висновок:** Похідна функції показує швидкість зростання оцінки в залежності від кількості годин навчання. Так як похідна є сталою та дорівнює 10, ми спостерігаємо постійний приріст оцінки на 10 балів з кожною додатковою годиною навчання.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



$ f(x) = \sin(x) $

```python
x = sp.symbols('x')

func_sin = sp.sin(x)
derivative_sin = sp.diff(func_sin, x)

print("Похідна лінійної функції sin(x):", derivative_sin)
```

```python
xs = np.linspace(0,10,100)
func_sin = np.sin(xs)
dfunc_sin = np.cos(xs)

plt.figure(figsize=(10, 5))
plt.plot(xs, func_sin, label = 'f(x) = sin(x)', color='blue')
plt.plot(xs, dfunc_sin, label="f'(x) = cos(x)", color='green')

plt.title("Графік функції $sin(x)$ та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

$ f(x) = \ln(x) $

```python
x = sp.symbols('x')

func_log = sp.log(x)
derivative_log = sp.diff(func_log, x)

print("Похідна лінійної функції ln(x):", derivative_log)
```

```python
xs = np.linspace(0.01,10,100)
func_log = np.log(xs)
dfunc_log = 1/xs

plt.figure(figsize=(10, 5))
plt.plot(xs, func_log, label = 'f(x) = ln(x)', color='blue')
plt.plot(xs, dfunc_log, label="f'(x) = 1/x", color='green')

plt.title("Графік функції $ln(x)$ та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.ylim(-5, 10)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')

func_costs = x**2 - 5*x + 1
derivative_costs = sp.diff(func_costs, x)

print("Похідна функції x**2 - 5*x + 1:", derivative_costs)
```

```python
xs = np.linspace(0,100,500)
func_costs = xs**2 - 5*xs + 1
dfunc_costs = 2*xs - 5

plt.figure(figsize=(10, 5))
plt.plot(xs, func_costs, label = r'$f(x) = x^2 - 5x + 1$', color='blue')
plt.plot(xs, dfunc_costs, label="f'(x) = 2x - 5", color='green')

plt.title("Графік функції $x^2 - 5x + 1$ та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.ylim(-10, 100)
plt.tight_layout()

plt.show()
```
